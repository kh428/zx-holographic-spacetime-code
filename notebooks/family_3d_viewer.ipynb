# Bond-extended spacetime blocks — interactive 3D viewer

Builds the $n=0,1,2$ blocks at depth $K$ and renders them with the [pyzx 3D viewer](https://github.com/kh428/pyzx_3d_viewer). Pauli webs can be overlaid (red/green/blue = X/Z/Y).

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import numpy as np
from spacetime.rim_basis import rim_basis_code
from spacetime.core import foliated_general as FG
from spacetime import e45, zx3d
from spacetime import wan_view as WV
print('ready')
```

```python
K = 2
FAM = {}
for n in (0, 1, 2):
    code2 = rim_basis_code(n)
    nq = code2['sx'].shape[1]
    blk = FG.build_general_foliated_block(code2['sx'], code2['sz'], K)
    g, vm = FG.to_zx(blk)
    P = e45.patch_layers(n)
    txy = np.asarray(P['xy'], float)
    xyG = np.zeros((nq, 2))
    for i in range(P['T']):
        c = txy[i]
        base = np.arctan2(c[1], c[0]) if np.hypot(*c) > 1e-9 else 0.0
        for s_ in range(4):
            th = base + 2*np.pi*(s_ + 0.5)/4
            xyG[4*i + s_] = c + 0.14*np.array([np.cos(th), np.sin(th)])
    WV.embed_foliated(g, vm, blk, xyG)
    FAM[n] = g
    print(f'n={n}: {g.num_vertices()} vertices')
```

```python
# drag to orbit, scroll to zoom (needs internet once for the import map)
n = 0
zx3d.draw_graph_3d_pw(FAM[n], None, node_size=0.14, edge_cyli_radius=0.05)
```

```python
from pyzx.web import compute_stabilisers
n = 1
webs = compute_stabilisers(FAM[n])
pick = max(webs, key=lambda w: len(list(w.half_edges())))
zx3d.draw_graph_3d_pw(FAM[n], [pick], node_size=0.10, edge_cyli_radius=0.035)
```
